Part 2 - Tuples

```python
ages = (20, 25, 22, 30, 28)
```

```python
# 1.print the tuple
print("tuple",ages)
```

```python
#2.first and last element
print("First element",ages[0])
print("Last element",ages[-1])
```

```python
#3.display: the maximum valum, the minimum value, the number of elements
print("Maximum valum",max(ages))
print("Minimum valum",min(ages))
print("Number of elements",len(ages))
```
